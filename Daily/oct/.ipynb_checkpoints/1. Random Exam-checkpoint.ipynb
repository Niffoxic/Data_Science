### 11. Consider a classifier that generates the output as below. The classifier will decide the label based on the threshold. If the output is larger or equal to the threshold, the predicted label will be 'Yes'; otherwise, 'No'

```python
# yes = True, no = False
data = [
    [0.95, True],
    [0.93, True],
    [0.87, False],
    [0.85, False],
    [0.85, True],
    [0.85, False],
    [0.76, True],
    [0.53, True],
    [0.43, False],
    [0.25, True],
]
```

### If we set the threshold at 0.70, what is the accuracy of the classifier:

```python
def calculate_accuracy(threshold=0.70):
    corrects = 0.0
    for i in data:
        corrects += (i[1] == (i[0] >= threshold))
        # print(i[0], i[1], corrects)
    return corrects / len(data)
```

**Answer:**

```python
calculate_accuracy()
```

### 12. Same question but to find classifiers

```python
def classifiers(threshold=0.70):
    tp, tn, fp, fn = 0, 0, 0, 0
    for i in data:
        if i[1] == True and (i[0] >= threshold):
            tp += 1
        elif not i[1] and not (i[0] >= threshold):
            tn += 1
        elif not i[1] and (i[0] >= threshold):
            fp += 1
        else:
            fn += 1
        # print(f"TP: {tp}, FN: {fn}, FP: {fp}, FN: {fn} ")
    return [tp, fn, fp, tn]
```

### Answer: 

```python
classifiers()
```

### 13. Same question but with threshold = 0.50

**Answer:**

### 14. Same question with same threshold but have to find TP, FN, FP, Tn

### 15. same question but to find F-measure

```python
def calculate_score(threshold=0.50):
    tp, fn, fp, tn = classifiers(threshold)
    
    precision = tp / (tp + fp)
    recall = tp / (tp + fn)
    
    return (2 * (precision * recall)) / (precision + recall)
```

### Answer:

```python
calculate_score()
```

### 18. Please calculate the accuracy of the classification, and the precision, recall, and f1 score of each class from the confusion matrix, then discuss their difference in their suitable their usage.

```python
import numpy as np

confustion_matrix = [
    [15, 1, 3, 5],
    [1, 9, 2, 0],
    [5, 2, 20, 0],
    [0, 4, 1, 10]
]
```

```python
def calculate_acc(index):
    true, overall = 0.0, 0.0
    for i in range(len(confustion_matrix)):
        if i == index:
            true = confustion_matrix[index][i]
        overall += confustion_matrix[index][i]
    return true / overall
```

```python
def calculate_each_scores():
    precision = np.diag(confustion_matrix) / np.sum(confustion_matrix, axis=0)
    recall = np.diag(confustion_matrix) / np.sum(confustion_matrix, axis=1)
    
    labels = ["Apple", "Banana", "Watermelon", "Strawberry"]
    scores = {}
    for i, key in enumerate(labels):
        scores[key] = {
            "Accuracy": calculate_acc(i),
            "Precision": precision[i],
            "Recall": recall[i],
            "F1-score": (2 * (precision[i] * recall[i]) / (precision[i] + recall[i]))
        }
    return scores
```

```python
res = calculate_each_scores()
for key, val in res.items():
    print(f"{key}:\n {val}\n\n")
```
