```python
def normal(x, mu, sigma):
    p = 1 / math.sqrt(2 * math.pi * sigma ** 2)
    return p * np.exp(-0.5 * (x - mu) ** 2 / sigma ** 2)
```

```python
x = np.arange(-7, 7, 0.01)
params = [(0, 1), (0, 2), (3, 1)]
d2l.plot(x, [normal(x, mu, sigma=sigma) for mu, sigma in params], xlabel="x",
         ylabel="p{X}", figsize=(4.5, 2.5),
         legend=[f'mean {mu}, std {sigma}' for mu, sigma in params])
```

```python
def add_to_class(Class):
    def wrapper(obj):
        setattr(Class, obj.__name__, obj)
    return wrapper
```

```python
class A:
    def __init__(self):
        self.b = 1
a = A()
```

```python
a.b
```

```python
@add_to_class(A)
def do(self):
    print("Class attrib B is:", self.b)
```

```python
a.do()
```

```python
class HyperParameters:
    def save_hyperparameters(self, ignore=[]):
        raise NotImplememented
```

```python
test = HyperParameters()
test.save_hyperparameters()
```

```python
class B(d2l.HyperParameters):
    def __init__(self, a, b, c):
        self.save_hyperparameters(ignore=["c"])
        print("a=", self.a, " b=", self.b)
        print("there is no c=", not hasattr(self, "c"))
```

```python
b = B(1, 2, 3)
```

```python
class ProgressBoard(d2l.HyperParameters):
    def __init__(self, xlabel=None, ylabel=None, xlim=None,
                 ylim=None, xscale="linear", yscale="linear",
                 ls=["-", "--", "-.", ":"], colors=["C0", "C1", "C2", "C3"],
                 fig=None, axes=None, figsize=(3.5, 2.5), display=True):
        self.save_hyperparameters()
    
    def draw(self, x, y, label, every_n=1):
        raise NotImplemented
```

```python
board = d2l.ProgressBoard("x")
for x in np.arange(0, 10, 0.1):
    board.draw(x, np.sin(x), "sin", every_n=2)
    board.draw(x, np.sin(x), "cos", every_n=10)
```
