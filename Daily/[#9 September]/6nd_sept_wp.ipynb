```python
import pandas as pd
```

```python
data = pd.read_csv("diabetes_data_upload.csv")
```

```python
data.head(10)
```

```python
from sklearn import preprocessing
```

```python
label_encoder = preprocessing.LabelEncoder()
```

```python
for i in data:
  if i == "Age": continue
  data[i] = label_encoder.fit_transform(data[i])
```

```python
data.head()
```

```python
data = data.astype(float)
```

```python
data.head()
```

```python
data["Age"] = data["Age"] / max(data["Age"])
```

```python
data.head()
```

```python
X, Y = data.drop(["class"], axis=1), data["class"]
```

```python
X.head()
```

```python
Y.head()
```

```python
Y.shape
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
```

```python
logistics = LogisticRegression()
```

```python
from sklearn.model_selection import train_test_split
```

```python
train_data, test_data, train_labels, test_labels = train_test_split(X, Y, test_size=0.2, shuffle=True)
```

```python
his = logistics.fit(train_data, train_labels)
```

```python
preds = logistics.predict(test_data)
```

```python
score = logistics.score(test_data, test_labels)
score
```

```python
from sklearn import metrics
import seaborn as sns
```

```python
cm = metrics.confusion_matrix(test_labels, preds)
print(cm)
```

```python
plt.figure(figsize=(9,9))
sns.heatmap(cm, annot=True, fmt=".3f", linewidths=.5, square = True, cmap = 'Blues_r');
plt.ylabel('Actual label');
plt.xlabel('Predicted label');
all_sample_title = 'Accuracy Score: {0}'.format(score)
plt.title(all_sample_title, size = 15);
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score
```

```python
print("precision_score:", precision_score(test_labels, preds))
```

```python
print("recall_score:", recall_score(test_labels, preds))
```

```python
print("f1_score:", f1_score(test_labels, preds))
```

```python
def build_model(test_size):
  train_data, test_data, train_labels, test_labels = train_test_split(X, Y, test_size=test_size, shuffle=True)
  model = LogisticRegression()
  model.fit(train_data, train_labels)
  preds = model.predict(test_data)
  res = {
      "precision_score": precision_score(test_labels, preds),
      "recall_score": recall_score(test_labels, preds),
      "f1_score":  f1_score(test_labels, preds),
      "accuracy": model.score(test_data, test_labels)
  }
  return res
```

```python
ts = [.4, .3, .2, .1]
models = []
for i in range(4):
  models.append(build_model(ts[i]))
```

```python
models
```

```python
vals  = [[("precision_score", [i["precision_score"] for i in models]),
      ("recall_score", [i["recall_score"] for i in models])],
      [("f1_score", [i["f1_score"] for i in models]),
      ("accuracy", [i["accuracy"] for i in models])]]
vals
```

```python
vals[0][1]
```

```python
mn = [1, 2, 3, 4]
mn_names = ["60-40 split", "70-30 split", "80-20 split", "90-10 split"]
```

```python
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

for i in range(2):
  for j in range(2):
    axes[i][j].scatter(mn, vals[i][j][1], color="orange")
    axes[i][j].plot(mn, vals[i][j][1])
    axes[i][j].set_title(vals[i][j][0], fontdict={"fontsize": 24})
    axes[i][j].set_xlabel("Models", fontdict={"fontsize": 18})
    axes[i][j].set_ylabel("Scores", fontdict={"fontsize": 18})
    axes[i][j].grid(True)
    axes[i][j].grid(color="b", ls="-", lw=0.25)
    for k, txt in enumerate(mn_names):
      axes[i][j].annotate(txt, (mn[k], vals[i][j][1][k]), fontsize=16)
fig.tight_layout()
plt.show()
```
