# Linear Regression with cuML

## Python Simple Linear Regression Code.
``Limited to one feature``

```python
import numpy as np
import pandas as pd
import plotly.express as px
```

```python
df = pd.read_csv("data/advertising.csv")
df.head()
```

```python
df["total_spend"] = df.TV + df.Radio + df.Newspaper
df.head()
```

```python
fig = px.scatter(df, x="total_spend", y="Sales",
                hover_data=["Sales"],
                trendline="ols",
                title="Advertising Data",
                labels={
                    "Sales": "Sales (%)",
                    "total_spend": "Total Amount Spend ($)"
                })


fig.update_layout(xaxis_range=[0, max(df.total_spend) + 10],
                  yaxis_range=[1, max(df.Sales) + 3],
                 title_x=0.5,
                 font=dict(size=18))
fig.data[1].update(line_color="red")
fig.show()
```

```python
np.polyfit(df.total_spend, df.Sales, deg=1)
# b1x + b0
```

```python
# b3x ** 3 + b2 ** 2 + b1x ** b0
w = np.polyfit(df.total_spend, df.Sales, deg=3)
w
```

```python
sample_x = df.total_spend[0]
sample_y = df.Sales[0]
sample_x, sample_y
```

```python
sample_yh = sum(w[:-1] * sample_x) + w[-1]
sample_yh, sample_y # error is huge xD ML
```
