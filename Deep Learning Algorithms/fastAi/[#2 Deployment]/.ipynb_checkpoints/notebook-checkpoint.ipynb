```python
animes = ["one piece anime", "dragon ball anime", "demon slayer anime"]
path = Path("data")
```

```python
for anime in animes:
    dest = (path/anime)
    dest.mkdir(exist_ok=True)
    results = search_images_ddg(f"{anime} images")
    download_images(dest, urls=results)
```

```python
fns = get_image_files(path)
fns
```

```python
failed = verify_images(fns)
failed
```

```python
len(failed) # damn
```

```python
failed.map(Path.unlink)
```

```python
data_block = DataBlock(
    blocks=(ImageBlock, CategoryBlock),
    get_items=get_image_files,
    splitter=RandomSplitter(valid_pct=0.2, seed=42),
    get_y=parent_label,
    item_tfms=Resize(128)
)
```

```python
dls = data_block.dataloaders(path)
```

```python
dls.train.show_batch(max_n=12, nrows=3)
```

```python
learn = vision_learner(dls, resnet50, metrics=error_rate)
learn.fine_tune(10)
```

```python
interp = ClassificationInterpretation.from_learner(learn)
interp.plot_confusion_matrix() # damn didnot know about data XD
```

```python
interp.plot_top_losses(4, nrows=2, figsize=(17, 4))
```
